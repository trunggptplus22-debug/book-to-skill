# Document IR and technical relevance classifier
The first prototype accepts text already extracted by the repository parser. It retains provenance placeholders explicitly and does not claim layout-perfect page or table recovery.

```python
from pathlib import Path
import sys, json
sys.path.insert(0, str(Path.cwd()))
from astro_technical_extraction.document_ir import DocumentIR
from astro_technical_extraction.classifier import RuleBasedClassifier
from astro_technical_extraction.coverage import build_coverage_report
sample = """The Ascendant is calculated from local sidereal time and geographic latitude.
Subtract the longitude of the Sun from the longitude of the Moon.
The author was born in the sixteenth century.
The soul participates in the divine order of the cosmos.
In this example chart, Mars is the significator of the tenth house.
A planet in domicile has essential dignity."""
document = DocumentIR.from_text("SAMPLE", "synthetic_fixture.txt", sample)
classifier = RuleBasedClassifier()
classifications = classifier.classify_many(document.blocks)
for item in classifications:
    print(json.dumps(item.to_dict(), ensure_ascii=False))
print(json.dumps(build_coverage_report(document, classifications), ensure_ascii=False, indent=2))
```
