# Document IR and technical classifier
This pilot accepts text extracted by the repository parser. It does not claim table/layout fidelity.

```python
import json,sys
from pathlib import Path
sys.path.insert(0,str(Path.cwd()))
from astro_technical_extraction.document_ir import DocumentIR
from astro_technical_extraction.classifier import RuleBasedClassifier
from astro_technical_extraction.coverage import build_coverage_report
fixture='''Calculate the Ascendant from local sidereal time.\nThe author was born in the sixteenth century.\nThe soul participates in the divine order of the cosmos.\nIn this example chart Mars signifies the tenth house.\nA planet in domicile has essential dignity.'''
doc=DocumentIR.from_text('FIXTURE','synthetic.txt',fixture)
clf=RuleBasedClassifier()
results=clf.classify_many(doc.blocks)
for result in results: print(json.dumps(result.to_dict(),ensure_ascii=False))
print(json.dumps(build_coverage_report(doc,results),ensure_ascii=False,indent=2))
```
