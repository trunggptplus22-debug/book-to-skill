## Source manifest
Set the source path only when the uploaded file is exposed to this runtime. Do not fabricate paths or page metadata.

```python
SOURCE_ID = "SRC001"
SOURCE_PATH = None  # Replace with a real mounted path after checking available files.
if SOURCE_PATH is not None:
    p = Path(SOURCE_PATH)
    if not p.is_file():
        raise FileNotFoundError(p)
    import hashlib
    digest = hashlib.sha256(p.read_bytes()).hexdigest()
    print({"source_id": SOURCE_ID, "file_name": p.name, "size_bytes": p.stat().st_size, "sha256": digest})
else:
    print("No source file path configured; no source processing performed.")
```
