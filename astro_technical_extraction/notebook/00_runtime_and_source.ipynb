## Source file check
Set SOURCE_PATH only after verifying the uploaded file is visible in this runtime.

```python
SOURCE_ID='SRC001'
SOURCE_PATH=None
if SOURCE_PATH is None:
    print('No source path configured; no source processing performed.')
else:
    p=Path(SOURCE_PATH)
    if not p.is_file(): raise FileNotFoundError(p)
    import hashlib
    print({'source_id':SOURCE_ID,'file_name':p.name,'size_bytes':p.stat().st_size,'sha256':hashlib.sha256(p.read_bytes()).hexdigest()})
```
