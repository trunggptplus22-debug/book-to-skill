# End-to-end technical candidate extraction
This synthetic fixture demonstrates pipeline mechanics, not astrology accuracy. All extracted records require review.

```python
from astro_technical_extraction import DocumentIR,RuleBasedClassifier,build_coverage_report,extract_records,validate_records,build_dependency_graph,export_bundle,export_json
from astro_technical_extraction.benchmark import evaluate
fixture='''Calculate the Ascendant from local sidereal time.
Formula: x = sin(a) + cos(b).
Table: traditional dignities and bounds.
Worked example chart: judge Mars by house and reception.
This historical passage describes a manuscript.'''
doc=DocumentIR.from_text('SYNTHETIC','fixture.txt',fixture)
classifications=RuleBasedClassifier().classify_many(doc.blocks)
coverage=build_coverage_report(doc,classifications)
records=extract_records(doc,classifications)
validation=validate_records(doc,records,coverage)
graph=build_dependency_graph(records)
bundle=export_bundle(doc,classifications,records,coverage,validation,graph)
print('Coverage:',coverage)
print('Validation:',validation)
print(export_json(bundle))
print('Example benchmark:',evaluate({'a':'INCLUDE','b':'EXCLUDE'},{'a':'INCLUDE','b':'REVIEW'}))
```
